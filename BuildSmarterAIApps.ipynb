**JSON PARSER**

```python
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field

class Joke(BaseModel):
  setup: str = Field(description="question to set up a joke")
  punchline: str = Field(description="answer to resolve the joke")
```

```python
joke_query = "tell me a joke"
output_parser = JsonOutputParser(pydantic_object=Joke)
format_instructions = output_parser.get_format_instructions()

prompt = PromptTemplate(
    template="Answer the user query.\n{format_instructions}\n{query}\n",
    input_variables=["query"],  # Dynamic variables that will be provided when invoking the chain
    partial_variables={"format_instructions": format_instructions},  # Static variables set once when creating the prompt
)

chain = prompt | llama_llm | output_parser

chain.invoke({"query": joke_query})
```

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser

output_parser = CommaSeparatedListOutputParser()
format_instructions = output_parser.get_format_instructions()

prompt = PromptTemplate(
    template="List five {subject}.\n{format_instructions}",
    input_variables=["subject"],
    partial_variables={"format_instructions": format_instructions}
)


chain = prompt | llama_llm | output_parser

response = chain.invoke({"subject":"top 5 tamil movies in 2026"})
print(response)
```

## Retriever

```python
# Use the docsearch vector store as a retriever
# This converts the vector store into a retriever interface that can fetch relevant documents
retriever = docsearch.as_retriever()

# Invoke the retriever with the query "Langchain"
# This will:
# 1. Convert the query text "Langchain" into an embedding vector
# 2. Perform a similarity search in the vector store using this embedding
# 3. Return the most semantically similar documents to the query
docs = retriever.invoke("Langchain")

# Access the first (most relevant) document from the retrieval results
# This returns the full Document object including:
# - page_content: The text content of the document
# - metadata: Any associated metadata like source, page numbers, etc.
# The returned document is the one most semantically similar to "Langchain"
docs[0]
```

##Parent Document Retriever

It’s a retrieval trick for RAG, the parent document retriever. It solves a tension in chunking:

Small chunks embed well, because the vector captures one focused idea, so search is accurate.
Large chunks give the LLM enough surrounding context to answer properly.

The retriever uses both. It splits each document into big parent chunks (2000 chars) and splits each parent into small child chunks (400 chars). Only the children are embedded and stored in Chroma. The parents are kept in a plain in-memory store. At query time, it finds the best-matching child chunks, then returns the larger parent each one came from.

```python
from langchain_chroma import Chroma
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.stores import InMemoryStore
from langchain_classic.retrievers import ParentDocumentRetriever
```

```python
parent_splitter = RecursiveCharacterTextSplitter(chunk_size=2000, chunk_overlap=20)
child_splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=20)

vectorstore = Chroma(
    collection_name="split_parents",
    embedding_function=embedding

)

store = InMemoryStore()

retriever = ParentDocumentRetriever(
    vectorstore=vectorstore,
    docstore=store,
    child_splitter=child_splitter,
    parent_splitter=parent_splitter,
)

retriever.add_documents(document)
```

```python
print(len(list(store.yield_keys())))   # number of parent chunks stored

# Child chunks (small, from the vector store)
sub_docs = vectorstore.similarity_search("Langchain")
print(sub_docs[0].page_content)
print("#"*7)
# Parent chunk (large, what the retriever returns)
retrieved_docs = retriever.invoke("Langchain")
print(retrieved_docs[0].page_content)
```

```python
query = "Langchain"
children = vectorstore.similarity_search(query, k=4)

for c in children:
    pid = c.metadata["doc_id"]
    parent = store.mget([pid])[0]
    print(f"child {c.id}  ->  parent {pid}")
    print("  matched:", c.page_content[:70].replace("\n", " "))
    print("  parent length:", len(parent.page_content), "chars\n")
```

```python
from langchain_anthropic import ChatAnthropic
from langchain_classic.chains import RetrievalQA

llm = ChatAnthropic(
    model = "claude-haiku-5-5",
    api_key = userdata.get('ANTHROPIC_API_KEY')
)

qa = RetrievalQA.from_chain_type(
    llm = llm,
    chain_type = "stuff",
    retriever = docsearch.as_retriever(),
    return_source_documents = True
)

result = qa.invoke({"query": "what is this paper discussing?"})
print(result["result"])
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

prompt = ChatPromptTemplate.from_template(
    "Answer using only the context below.\n\nContext:\n{context}\n\nQuestion: {question}"
)

def format_docs(docs):
    return "\n\n".join(d.page_content for d in docs)

chain = (
    {"context": docsearch.as_retriever() | format_docs, "question": RunnablePassthrough()}
    | prompt | llm | StrOutputParser()
)

print(chain.invoke("what is this paper discussing?"))
```

```python
import os
os.environ["USER_AGENT"] = "rag-demo"             # silences a WebBaseLoader warning
os.environ["ANONYMIZED_TELEMETRY"] = "False"      # silences Chroma telemetry noise
# os.environ["ANTHROPIC_API_KEY"] = "sk-ant-..."  # only needed for step 8

from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

# 1. Load a document
loader = WebBaseLoader("https://python.langchain.com/docs/introduction/")
documents = loader.load()

# 2. Split into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=20)
chunks = text_splitter.split_documents(documents)

# 3. Embedding model (local, free; replaces WatsonxEmbeddings)
embedding = HuggingFaceEmbeddings(
    model_name="ibm-granite/granite-embedding-278m-multilingual"
    # lighter option: "sentence-transformers/all-MiniLM-L6-v2"
)

# 4. Vector store
vector_store = Chroma.from_documents(chunks, embedding)

# 5. Retriever
retriever = vector_store.as_retriever(search_kwargs={"k": 3})

# 6. Search function
def search_documents(query, top_k=3):
    """Search for documents relevant to a query"""
    docs = retriever.invoke(query)      # get_relevant_documents is deprecated
    return docs[:top_k]

# 7. Test queries
test_queries = [
    "What is LangChain?",
    "How do retrievers work?",
    "Why is document splitting important?",
]

for query in test_queries:
    print(f"\nQuery: {query}")
    results = search_documents(query)
    print(f"Found {len(results)} relevant documents:")
    for i, doc in enumerate(results, 1):
        print(f"\nResult {i}: {doc.page_content[:150].strip()}...")
        print(f"Source: {doc.metadata.get('source', 'Unknown')}")
```
